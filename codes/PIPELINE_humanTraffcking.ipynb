```python
dataset_id = 'ijzp-q8t2'
endpoint = 'data.cityofchicago.org/'
first_filter_field= 'primary_type'
first_filter_value= 'HUMAN TRAFFICKING'

response = get_data(dataset_id = dataset_id, 
                    endpoint = endpoint,
                    first_filter_field =first_filter_field, 
                    first_filter_value = first_filter_value
                    )
```

```python
response
```

```python
import pandas as pd

df = pd.DataFrame(response)
df.info()
```

```python
df['dateOnly'] = pd.to_datetime(df.date).dt.date

df['yearMonth'] = pd.to_datetime(df.date).dt.strftime('%Y-%m')
```

```python
df.drop_duplicates('case_number')\
  .query('year >= "2024"')\
  .yearMonth.value_counts()\
  .sort_index()
```
